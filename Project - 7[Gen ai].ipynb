```python
# Import text-generation pipeline
from transformers import pipeline

# Load GPT-2 using PyTorch
generator = pipeline(
    "text-generation",
    model="gpt2",
    framework="pt"
)

print("GPT-2 loaded successfully!")
```

```python
# Get user input
prompt = input("Enter a sentence or topic: ")

# Generate text
result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

# Display generated text
print("\nGenerated Text:")
print(result[0]["generated_text"])
```

```python
# Test three different inputs
inputs = [
    "Artificial Intelligence is",
    "The future of technology is",
    "Once upon a time"
]

# Generate text for each input
for i, prompt in enumerate(inputs, 1):
    result = generator(
        prompt,
        max_new_tokens=30,
        num_return_sequences=1
    )

    print(f"\nInput {i}: {prompt}")
    print("Output:", result[0]["generated_text"])
```
